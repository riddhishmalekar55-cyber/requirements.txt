```python
import os
from google.colab import files
import pandas as pd

file_name = 'numeric_only.csv'

# Check if the file exists in the current directory
if not os.path.exists(file_name):
    print(f"File '{file_name}' not found. Please upload it.")
    # Prompt user to upload the file
    uploaded = files.upload()
    if file_name in uploaded:
        print(f"'{file_name}' uploaded successfully.")
        # File should now be in the current working directory, so read it
        df = pd.read_csv(file_name)
    else:
        print(f"Failed to upload '{file_name}'. Please ensure you selected the correct file.")
        df = None # Set df to None or handle error as appropriate
else:
    print(f"File '{file_name}' found.")
    df = pd.read_csv(file_name)
```

```python
df= df[351:400]
```

```python
df.info()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logmodel = LogisticRegression()
```

```python
from sklearn.model_selection import train_test_split

# Create a binary target variable for classification
df['High_Spender'] = df['Yearly Amount Spent'].apply(lambda x: 1 if x > df['Yearly Amount Spent'].median() else 0)

# Define features (X) and target (y)
X = df[['Avg. Session Length', 'Time on App', 'Time on Website', 'Length of Membership']]
y = df['High_Spender']

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)

# Train the logistic regression model
logmodel.fit(X_train, y_train)

# Make predictions on the test set
predictions = logmodel.predict(X_test)

print("Data prepared, model trained, and predictions made. You can now run the classification report.")
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,predictions))
```
